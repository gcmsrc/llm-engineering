To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display
import os
```

```python
# constants}
MODEL_GPT = 'gpt-4o-mini'
MODEL_OLLAMA = 'phi3'
```

```python
# set up environment
load_dotenv()
if os.getenv('OPENAI_API_KEY') is None:
    raise ValueError("OPENAI_API_KEY is not set")
else:
    print("OPENAI_API_KEY is set")
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
system_prompt = """You are an expert software engineer.
You provide answers to a question by using analogy from the biology world.
This shows your deep understanding of the subject as you can explain complex concepts in a simple way.
"""

user_prompt_prefix = """Can you help me with the following question?

"""
```

```python
def format_messages(question):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + question}
    ]
```

```python
def answer_with_ai(question, local=False, model=MODEL_GPT):
    messages = format_messages(question)
    client = OpenAI() if not local else OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
    try:
        response = client.chat.completions.create(
            model=model,
            messages=messages
        )
    except Exception as e:
        print(f"Error: {e}")
        return None
    
    return response


    
```

```python
r = answer_with_ai(question)
```

```python
r_ollama = answer_with_ai(question, local=True, model=MODEL_OLLAMA)
```

```python
def display_response(response):
    if response is None:
        Markdown("No response from the AI")
        return
    
    display(Markdown(response.choices[0].message.content))       
```

```python
display_response(r)
```

```python
display_response(r_ollama)
```
